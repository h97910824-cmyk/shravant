```python
import pandas as pd
```

```python
df = pd.read_csv('Auto_mpg.csv', usecols=['mpg', 'horsepower', 'weight'])
print(df)
```

```python
df['std-mpg'] = (df['mpg'] - df['mpg'].mean()) / df['mpg'].std()
df['horsepower'] = pd.to_numeric(df['horsepower'],errors='coerce')
df['std-horsepower'] = (df['horsepower'] - df['horsepower'].mean()) / df['horsepower'].std()
df['std-weight'] = (df['weight'] - df['weight'].mean()) / df['weight'].std()
print(df)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[['mpg-std', 'horsepower-std', 'weight-std']] = scaler.fit_transform(df[['mpg', 'horsepower', 'weight']])
df[['mpg', 'mpg-std', 'horsepower', 'horsepower-std', 'weight', 'weight-std']]
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
df = pd.read_csv("Auto_mpg.csv")
data = df[["origin"]]
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(data)
encoded_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out(["origin"]),index=df.index)
df1 = pd.concat([df, encoded_df], axis=1)
print(df1)
```
